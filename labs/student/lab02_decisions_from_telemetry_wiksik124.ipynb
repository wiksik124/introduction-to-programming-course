## 8. Implement the telemetry classifier

Use the four input variables below and an `if`/`elif`/`else` chain. Assign exactly one of these strings to `status`:

```text
CRITICAL
SAFE MODE
WARNING
NOMINAL
```

Do not start coding until your decision table is complete.

````python
battery_charge = 15
bus_voltage_v = 24
panel_temperature_c = -20
comms_ok = True

# Implement the priority rules here.
status = "NOT IMPLEMENTED"

if battery_charge < 15 or bus_voltage_v < 22:
    status = "CRITICAL"
elif battery_charge < 25 or bus_voltage_v < 24 or panel_temperature_c < -20 or panel_temperature_c > 85 or comms_ok == False:
    status = "SAFE MODE"
elif battery_charge <= 35 or panel_temperature_c > 75:
    status = "WARNING"
else:
    status = "NOMINAL"

print("Spacecraft status:", status)
````

### Explain your implementation

Why must the CRITICAL test appear before the SAFE MODE test? Why is `else` appropriate for NOMINAL?

**Your explanation:**

*Because python picks the first condition when it is met, so if the most critical test isn't met, it goes to the less critical one, in this example it skips to safe mode, then warning, and then nominal. If we put safe mode before critical, the critical condition would never execute, because the criteria for critical are also met in safe mode.*

*Else is appropriate for nominal, beceuse it is executed after every other status check fails. Nominal is the least severe condition and doesn't need any requirements to be met.*
